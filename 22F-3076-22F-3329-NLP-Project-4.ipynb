```python
# ============================================
# Cell 2: Setup Paths and Download Dataset
# ============================================
print("=" * 60)
print("STEP 2: Setting Up Paths")
print("=" * 60)

# Define paths - Dataset already uploaded in Kaggle notebook
dataset_path = "/kaggle/input/mimic-iv-ext-direct-1-0-0"  # Path to uploaded dataset
zip_file_path = "/kaggle/input/mimic-iv-ext-direct-1-0-0/mimic-iv-ext-direct-1.0.0.zip"  # Zip file path
extract_folder = "mimic_dataset_extracted"  # Jahan extract karenge

print(f"Dataset Input Path: {dataset_path}")
print(f"Zip File Path: {zip_file_path}")
print(f"Extract Folder: {extract_folder}")
print()

# Check if dataset path exists
if os.path.exists(dataset_path):
    print(f"✓ Dataset folder found!")
    print(f"Contents of {dataset_path}:")
    for item in os.listdir(dataset_path):
        print(f"  - {item}")
else:
    print(f"❌ Dataset folder not found at {dataset_path}")
    print("Please ensure the dataset is uploaded correctly.")

print()
```

```python
# ============================================
# Cell 3: Unzip the Dataset
# ============================================
print("=" * 60)
print("STEP 3: Unzipping the Dataset")
print("=" * 60)

# Create extraction folder if not exists
if not os.path.exists(extract_folder):
    os.makedirs(extract_folder)
    print(f"✓ Created folder: {extract_folder}")

# Unzip the file
try:
    # Check if zip file exists
    if not os.path.exists(zip_file_path):
        print(f"❌ Zip file not found at: {zip_file_path}")
        print("\nLet's check what files are available:")
        if os.path.exists(dataset_path):
            all_files = []
            for root, dirs, files in os.walk(dataset_path):
                for file in files:
                    all_files.append(os.path.join(root, file))
            
            print(f"Found {len(all_files)} files:")
            for f in all_files[:10]:  # Show first 10 files
                print(f"  - {f}")
            if len(all_files) > 10:
                print(f"  ... and {len(all_files) - 10} more files")
        
        # If files are already extracted, skip unzipping
        print("\n⚠️ Zip file not found. Checking if files are already extracted...")
        zip_file_path = None  # Mark as not available
        
    else:
        with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
            # Get list of files in zip
            file_list = zip_ref.namelist()
            print(f"Total files in zip: {len(file_list)}")
            print()
            
            # Extract all files
            print("Extracting files...")
            zip_ref.extractall(extract_folder)
            print(f"✓ Successfully extracted {len(file_list)} files to '{extract_folder}'")
        
except FileNotFoundError:
    print(f"❌ Error: File not found!")
    print("Files might already be in the dataset folder. Proceeding with direct access...")
except Exception as e:
    print(f"❌ Error during extraction: {str(e)}")
    print("Proceeding to check if files are already available...")

print()
```

```python
# ============================================
# Cell 4: List All Extracted Files
# ============================================
print("=" * 60)
print("STEP 4: Listing All Extracted Files")
print("=" * 60)

# Function to recursively find all files
def find_all_files(directory, extensions=['.json', '.jsonl']):
    """Find all files with specific extensions"""
    all_files = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            file_path = os.path.join(root, file)
            file_ext = os.path.splitext(file)[1].lower()
            if file_ext in extensions:
                all_files.append(file_path)
    return all_files

# Find all JSON/JSONL files - search in both extracted folder and input dataset
search_paths = []

# Add extracted folder if it exists
if os.path.exists(extract_folder):
    search_paths.append(extract_folder)

# Also search in the input dataset folder
if os.path.exists(dataset_path):
    search_paths.append(dataset_path)

json_files = []
for search_path in search_paths:
    json_files.extend(find_all_files(search_path, ['.json', '.jsonl']))

# Remove duplicates
json_files = list(set(json_files))

print(f"Total JSON/JSONL files found: {len(json_files)}")
print()

if json_files:
    print("List of files:")
    print("-" * 60)
    for idx, file_path in enumerate(json_files, 1):
        file_size = os.path.getsize(file_path) / (1024 * 1024)  # Size in MB
        print(f"{idx}. {file_path}")
        print(f"   Size: {file_size:.2f} MB")
else:
    print("❌ No JSON/JSONL files found!")

print()
```

```python
# ============================================
# Cell 5: Load and Verify Each File
# ============================================
print("=" * 60)
print("STEP 5: Loading and Verifying Files")
print("=" * 60)

file_info = []

for idx, file_path in enumerate(json_files, 1):
    print(f"\n[{idx}/{len(json_files)}] Processing: {os.path.basename(file_path)}")
    print("-" * 60)
    
    try:
        # Determine file type
        file_ext = os.path.splitext(file_path)[1].lower()
        
        # Load JSON file
        if file_ext == '.json':
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                
            # Check if it's a list or dict
            if isinstance(data, list):
                record_count = len(data)
                sample_record = data[0] if data else None
            elif isinstance(data, dict):
                record_count = 1
                sample_record = data
            else:
                record_count = 1
                sample_record = data
                
        # Load JSONL file (each line is a JSON object)
        elif file_ext == '.jsonl':
            data = []
            with open(file_path, 'r', encoding='utf-8') as f:
                for line in f:
                    if line.strip():
                        data.append(json.loads(line))
            record_count = len(data)
            sample_record = data[0] if data else None
        
        # Display info
       # print(f"✓ Successfully loaded!")
        # print(f"  Records: {record_count}")
        #print(f"  File size: {os.path.getsize(file_path) / 1024:.2f} KB")
        
        # Show sample record structure
        if sample_record:
            print(f"  Sample keys: {list(sample_record.keys()) if isinstance(sample_record, dict) else 'N/A'}")
            print(f"\n  Sample Record (first 500 chars):")
            print(f"  {str(sample_record)[:500]}...")
        
        # Store file info
        file_info.append({
            'file_name': os.path.basename(file_path),
            'file_path': file_path,
            'record_count': record_count,
            'file_size_kb': os.path.getsize(file_path) / 1024,
            'status': 'Success',
            'keys': list(sample_record.keys()) if isinstance(sample_record, dict) else []
        })
        
    except json.JSONDecodeError as e:
        print(f"❌ JSON Decode Error: {str(e)}")
        file_info.append({
            'file_name': os.path.basename(file_path),
            'file_path': file_path,
            'status': 'Failed - Invalid JSON',
            'error': str(e)
        })
    except Exception as e:
        print(f"❌ Error: {str(e)}")
        file_info.append({
            'file_name': os.path.basename(file_path),
            'file_path': file_path,
            'status': 'Failed',
            'error': str(e)
        })

print()
```

```python
# Cell 1: Import Required Libraries
print("=" * 60)
print("TASK 2: DATA PREPROCESSING")
print("=" * 60)
print()

import os
import json
import pandas as pd
import numpy as np
from pathlib import Path
import re
from collections import Counter
import warnings
warnings.filterwarnings('ignore')

# For text processing
NLTK_AVAILABLE = False
try:
    import nltk
    from nltk.tokenize import word_tokenize, sent_tokenize
    from nltk.corpus import stopwords
    
    # Try to download required NLTK data
    try:
        nltk.download('punkt', quiet=True)
        nltk.download('stopwords', quiet=True)
        nltk.download('punkt_tab', quiet=True)
        NLTK_AVAILABLE = True
        print("✓ NLTK libraries loaded successfully!")
    except:
        print("⚠️ NLTK download failed, using basic tokenization")
        NLTK_AVAILABLE = False
except ImportError:
    print("⚠️ NLTK not installed, using basic tokenization")
    NLTK_AVAILABLE = False

print("✓ All libraries imported successfully!")
print()
```

```python
# ============================================
# Cell 2: Load All JSON Files from Task 1
# ============================================
print("=" * 60)
print("STEP 1: Loading Dataset from Task 1")
print("=" * 60)

# Define paths
dataset_path = "/kaggle/input/mimic-iv-ext-direct-1-0-0/mimic-iv-ext-direct-1.0.0"

# Function to find all JSON files
def find_json_files(directory):
    json_files = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.json') or file.endswith('.jsonl'):
                json_files.append(os.path.join(root, file))
    return json_files

# Find all JSON files
json_files = find_json_files(dataset_path)
print(f"✓ Found {len(json_files)} JSON files")
print()

# Load all data into a list
all_records = []
file_sources = []

print("Loading files...")
for idx, file_path in enumerate(json_files, 1):
    if idx % 100 == 0:
        print(f"  Loaded {idx}/{len(json_files)} files...")
    
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            
        # Handle different data structures
        if isinstance(data, list):
            for record in data:
                record['source_file'] = os.path.basename(file_path)
                all_records.append(record)
        elif isinstance(data, dict):
            data['source_file'] = os.path.basename(file_path)
            all_records.append(data)
            
    except Exception as e:
        print(f"⚠️ Error loading {file_path}: {str(e)}")

print(f"\n✓ Successfully loaded {len(all_records)} total records")
print()
```

```python
# ============================================
# Cell 3: Initial Data Exploration
# ============================================
print("=" * 60)
print("STEP 2: Initial Data Exploration")
print("=" * 60)

# Convert to DataFrame for easier processing
df_original = pd.DataFrame(all_records)

print(f"Dataset Shape: {df_original.shape}")
print(f"Columns: {list(df_original.columns)}")
print()

# Show basic info
print("📊 Original Dataset Statistics:")
print("-" * 60)
print(f"Total Records: {len(df_original):,}")
print(f"Total Columns: {len(df_original.columns)}")
print(f"Memory Usage: {df_original.memory_usage(deep=True).sum() / (1024*1024):.2f} MB")
print()

# Check for missing values
print("Missing Values per Column:")
print("-" * 60)
missing_stats = df_original.isnull().sum()
missing_percent = (missing_stats / len(df_original) * 100).round(2)
missing_df = pd.DataFrame({
    'Missing Count': missing_stats,
    'Percentage': missing_percent
})
print(missing_df[missing_df['Missing Count'] > 0].to_string())
print()

# Show sample record
print("Sample Record (first record):")
print("-" * 60)
if len(df_original) > 0:
    print(json.dumps(df_original.iloc[0].to_dict(), indent=2, default=str)[:500] + "...")
print()
```

```python
# ============================================
# Cell 4: Data Cleaning - Remove Irrelevant Records
# ============================================
print("=" * 60)
print("STEP 3: Filtering Irrelevant Records")
print("=" * 60)

df_cleaned = df_original.copy()
initial_count = len(df_cleaned)

# Identify text fields (columns that contain clinical information)
text_columns = []
for col in df_cleaned.columns:
    if df_cleaned[col].dtype == 'object':
        sample_value = df_cleaned[col].dropna().iloc[0] if len(df_cleaned[col].dropna()) > 0 else ""
        if isinstance(sample_value, str) and len(str(sample_value)) > 20:
            text_columns.append(col)

print(f"Identified text columns: {text_columns}")
print()

# Filter 1: Remove records with all empty text fields
if text_columns:
    print("Filter 1: Removing records with empty text fields...")
    df_cleaned = df_cleaned.dropna(subset=text_columns, how='all')
    removed_1 = initial_count - len(df_cleaned)
    print(f"  Removed: {removed_1} records")
    print(f"  Remaining: {len(df_cleaned)} records")
    print()

# Filter 2: Remove very short records (less than 10 characters in main text)
print("Filter 2: Removing very short records...")
if text_columns:
    def is_valid_length(row):
        for col in text_columns:
            if pd.notna(row[col]) and len(str(row[col])) >= 10:
                return True
        return False
    
    df_cleaned = df_cleaned[df_cleaned.apply(is_valid_length, axis=1)]
    removed_2 = initial_count - removed_1 - len(df_cleaned)
    print(f"  Removed: {removed_2} records")
    print(f"  Remaining: {len(df_cleaned)} records")
    print()

# Filter 3: Remove duplicate records
print("Filter 3: Removing duplicate records...")
initial_dedup = len(df_cleaned)
df_cleaned = df_cleaned.drop_duplicates(subset=text_columns if text_columns else None, keep='first')
removed_3 = initial_dedup - len(df_cleaned)
print(f"  Removed: {removed_3} duplicate records")
print(f"  Remaining: {len(df_cleaned)} records")
print()

total_removed = initial_count - len(df_cleaned)
print(f"✓ Total records removed: {total_removed}")
print(f"✓ Clean records remaining: {len(df_cleaned)}")
print()
```

```python
# ============================================
# Cell 5: Handle Missing Values
# ============================================
print("=" * 60)
print("STEP 4: Handling Missing Values")
print("=" * 60)

# Check missing values in cleaned data
missing_after = df_cleaned.isnull().sum()
print("Missing values in cleaned dataset:")
print("-" * 60)
for col in df_cleaned.columns:
    missing_count = missing_after[col]
    if missing_count > 0:
        missing_pct = (missing_count / len(df_cleaned) * 100)
        print(f"{col}: {missing_count} ({missing_pct:.2f}%)")
print()

# Strategy 1: Fill numeric columns with median
numeric_cols = df_cleaned.select_dtypes(include=[np.number]).columns
if len(numeric_cols) > 0:
    print(f"Filling {len(numeric_cols)} numeric columns with median...")
    for col in numeric_cols:
        if df_cleaned[col].isnull().sum() > 0:
            median_val = df_cleaned[col].median()
            df_cleaned[col].fillna(median_val, inplace=True)
            print(f"  ✓ {col}: filled with {median_val}")
    print()

# Strategy 2: Fill text columns with empty string or 'Unknown'
text_cols = df_cleaned.select_dtypes(include=['object']).columns
if len(text_cols) > 0:
    print(f"Filling {len(text_cols)} text columns...")
    for col in text_cols:
        if df_cleaned[col].isnull().sum() > 0:
            # Don't fill main text columns, keep them for filtering
            if col not in text_columns:
                df_cleaned[col].fillna('Unknown', inplace=True)
                print(f"  ✓ {col}: filled with 'Unknown'")
    print()

# Final check
remaining_missing = df_cleaned.isnull().sum().sum()
print(f"✓ Remaining missing values: {remaining_missing}")
print()
```

```python
# ============================================
# Cell 6: Text Tokenization
# ============================================
print("=" * 60)
print("STEP 5: Text Tokenization")
print("=" * 60)

# Function to clean text
def clean_text(text):
    if pd.isna(text):
        return ""
    text = str(text)
    # Remove special characters but keep medical terms
    text = re.sub(r'[^\w\s\-\.]', ' ', text)
    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

# Function to tokenize text
def tokenize_text(text):
    if not text:
        return []
    
    # Try NLTK first if available
    if NLTK_AVAILABLE:
        try:
            tokens = word_tokenize(text.lower())
            return tokens
        except:
            pass
    
    # Fallback: Basic tokenization
    # Remove punctuation and split
    text = re.sub(r'[^\w\s]', ' ', text.lower())
    tokens = text.split()
    return tokens

# Add tokenized columns for main text fields
print("Tokenizing text fields...")
for col in text_columns[:3]:  # Tokenize first 3 text columns to save memory
    if col in df_cleaned.columns:
        print(f"  Processing column: {col}")
        # Clean text
        df_cleaned[f'{col}_cleaned'] = df_cleaned[col].apply(clean_text)
        # Tokenize
        df_cleaned[f'{col}_tokens'] = df_cleaned[f'{col}_cleaned'].apply(tokenize_text)
        # Token count
        df_cleaned[f'{col}_token_count'] = df_cleaned[f'{col}_tokens'].apply(len)
        print(f"    ✓ Avg tokens: {df_cleaned[f'{col}_token_count'].mean():.1f}")

print()
print("✓ Tokenization completed!")
print()

# Show tokenization statistics
print("Tokenization Statistics:")
print("-" * 60)
token_cols = [col for col in df_cleaned.columns if col.endswith('_token_count')]
for col in token_cols:
    print(f"{col}:")
    print(f"  Mean: {df_cleaned[col].mean():.1f}")
    print(f"  Median: {df_cleaned[col].median():.1f}")
    print(f"  Min: {df_cleaned[col].min()}")
    print(f"  Max: {df_cleaned[col].max()}")
print()
```

```python
# ============================================
# Cell 7: Create Combined Text Field for Embeddings
# ============================================
print("=" * 60)
print("STEP 6: Preparing Text for Embeddings")
print("=" * 60)

# Combine all text fields into one field for embedding
def combine_text_fields(row):
    combined = []
    for col in text_columns:
        if col in row and pd.notna(row[col]):
            combined.append(str(row[col]))
    return " ".join(combined)

print("Creating combined text field...")
df_cleaned['combined_text'] = df_cleaned.apply(combine_text_fields, axis=1)
df_cleaned['combined_text_cleaned'] = df_cleaned['combined_text'].apply(clean_text)
df_cleaned['combined_text_length'] = df_cleaned['combined_text_cleaned'].apply(len)

print("✓ Combined text field created!")
print()

print("Combined Text Statistics:")
print("-" * 60)
print(f"Mean length: {df_cleaned['combined_text_length'].mean():.1f} characters")
print(f"Median length: {df_cleaned['combined_text_length'].median():.1f} characters")
print(f"Min length: {df_cleaned['combined_text_length'].min()} characters")
print(f"Max length: {df_cleaned['combined_text_length'].max()} characters")
print()

# Show sample of cleaned text
print("Sample Cleaned Text:")
print("-" * 60)
if len(df_cleaned) > 0:
    sample_text = df_cleaned['combined_text_cleaned'].iloc[0]
    print(sample_text[:300] + "..." if len(sample_text) > 300 else sample_text)
print()
```

```python
# ============================================
# Cell 8: Generate Document IDs and Metadata
# ============================================
print("=" * 60)
print("STEP 7: Generating Document IDs and Metadata")
print("=" * 60)

# Create unique document IDs
df_cleaned['doc_id'] = ['DOC_' + str(i).zfill(6) for i in range(len(df_cleaned))]

# Create metadata dictionary for each document
def create_metadata(row):
    metadata = {
        'doc_id': row['doc_id'],
        'source_file': row.get('source_file', 'Unknown'),
        'text_length': row.get('combined_text_length', 0),
        'token_count': row.get(f'{text_columns[0]}_token_count', 0) if text_columns else 0
    }
    return metadata

df_cleaned['metadata'] = df_cleaned.apply(create_metadata, axis=1)

print("✓ Document IDs and metadata generated!")
print()

print(f"Total documents: {len(df_cleaned)}")
print(f"Sample document ID: {df_cleaned['doc_id'].iloc[0] if len(df_cleaned) > 0 else 'N/A'}")
print()
```

```python
# ============================================
# TASK 3: RETRIEVAL COMPONENT (NO INTERNET NEEDED)
# ============================================

print("=" * 60)
print("TASK 3: RETRIEVAL COMPONENT")
print("=" * 60)
print()

import os
import json
import pandas as pd
import numpy as np
from pathlib import Path
import pickle
import warnings
import re
import math
from collections import Counter
from tqdm import tqdm

warnings.filterwarnings('ignore')

print("✓ All base libraries loaded successfully!")
print()

# ============================================
# CUSTOM BM25 IMPLEMENTATION (No external lib needed)
# ============================================

class SimpleBM25:
    """Custom BM25 implementation - no external library needed"""
    
    def __init__(self, corpus, k1=1.5, b=0.75):
        """
        Initialize BM25
        corpus: list of tokenized documents
        k1: controls term frequency saturation (default: 1.5)
        b: controls how much effect document length has (default: 0.75)
        """
        self.corpus = corpus
        self.k1 = k1
        self.b = b
        
        # Calculate statistics
        self.doc_count = len(corpus)
        self.avg_doc_length = np.mean([len(doc) for doc in corpus])
        
        # Build IDF (Inverse Document Frequency)
        self.idf = self._calculate_idf()
    
    def _calculate_idf(self):
        """Calculate IDF for all terms"""
        idf = {}
        
        # Count documents containing each term
        for doc in self.corpus:
            for term in set(doc):
                if term not in idf:
                    idf[term] = 0
                idf[term] += 1
        
        # Convert to IDF scores
        for term in idf:
            doc_freq = idf[term]
            idf[term] = math.log(
                (self.doc_count - doc_freq + 0.5) / (doc_freq + 0.5) + 1
            )
        
        return idf
    
    def get_scores(self, query):
        """Calculate BM25 scores for query against all documents"""
        scores = np.zeros(self.doc_count)
        
        # Count query term frequencies
        query_freq = Counter(query)
        
        for doc_idx, doc in enumerate(self.corpus):
            doc_freq = Counter(doc)
            doc_length = len(doc)
            
            score = 0
            for term, query_count in query_freq.items():
                if term in doc_freq:
                    term_freq = doc_freq[term]
                    idf = self.idf.get(term, 0)
                    
                    # BM25 formula
                    numerator = idf * term_freq * (self.k1 + 1)
                    denominator = (
                        term_freq + 
                        self.k1 * (
                            1 - self.b + 
                            self.b * (doc_length / self.avg_doc_length)
                        )
                    )
                    
                    score += numerator / denominator
            
            scores[doc_idx] = score
        
        return scores

# ============================================
# Cell 1: Load Preprocessed Data from Task 2
# ============================================
print("=" * 60)
print("STEP 1: Loading Preprocessed Data")
print("=" * 60)

preprocessed_path = "preprocessed_data/preprocessed_documents.json"

if not os.path.exists(preprocessed_path):
    print(f"❌ Preprocessed data not found at: {preprocessed_path}")
    print("Please run Task 2 first!")
    raise FileNotFoundError("Preprocessed data missing")

with open(preprocessed_path, 'r', encoding='utf-8') as f:
    documents = json.load(f)

print(f"✓ Loaded {len(documents)} documents")
if documents:
    print(f"Sample document: {json.dumps(documents[0], indent=2)[:200]}...")
print()
```

```python
# ============================================
# Cell 2: Text Preprocessing
# ============================================
print("=" * 60)
print("STEP 2: Text Preprocessing for Retrieval")
print("=" * 60)

class TextPreprocessor:
    """Text preprocessor for clinical text"""
    
    def __init__(self):
        self.stop_words = {
            'the', 'a', 'an', 'and', 'or', 'but', 'in', 'on', 'at', 'to', 
            'for', 'of', 'with', 'by', 'from', 'as', 'is', 'was', 'are', 'were',
            'be', 'been', 'being', 'have', 'has', 'had', 'do', 'does', 'did',
            'will', 'would', 'could', 'should', 'may', 'might', 'can', 'must'
        }
    
    def clean_text(self, text):
        """Clean text"""
        if not text:
            return ""
        
        text = str(text).lower()
        text = re.sub(r'http\S+|www\S+', '', text)
        text = re.sub(r'[^a-z0-9\s\-\.]', ' ', text)
        text = re.sub(r'\s+', ' ', text)
        
        return text.strip()
    
    def tokenize(self, text):
        """Tokenize text"""
        if not text:
            return []
        
        tokens = text.split()
        tokens = [
            token for token in tokens 
            if token not in self.stop_words and len(token) > 2
        ]
        
        return tokens
    
    def process(self, text):
        """Complete pipeline"""
        cleaned = self.clean_text(text)
        tokens = self.tokenize(cleaned)
        return tokens

preprocessor = TextPreprocessor()
print("✓ Text preprocessor initialized!")
print()

# Preprocess documents
print("Preprocessing documents...")
preprocessed_texts = []
tokenized_docs = []

for doc in tqdm(documents, desc="Processing"):
    text = doc['text']
    cleaned = preprocessor.clean_text(text)
    tokens = preprocessor.tokenize(cleaned)
    
    preprocessed_texts.append(cleaned)
    tokenized_docs.append(tokens)

print(f"✓ Preprocessed {len(preprocessed_texts)} documents")
if tokenized_docs:
    print(f"Sample tokens: {tokenized_docs[0][:15]}")
print()
```

```python
# ============================================
# Cell 3: BM25 Retriever
# ============================================
print("=" * 60)
print("STEP 3: Building BM25 Sparse Retriever")
print("=" * 60)

class BM25Retriever:
    """BM25 retriever using custom implementation"""
    
    def __init__(self, documents, tokenized_docs):
        self.documents = documents
        self.tokenized_docs = tokenized_docs
        self.preprocessor = TextPreprocessor()
        
        print("Building BM25 index...")
        self.bm25 = SimpleBM25(tokenized_docs)
        print("✓ BM25 index built successfully!")
    
    def search(self, query, top_k=5):
        """Search using BM25"""
        query_tokens = self.preprocessor.tokenize(
            self.preprocessor.clean_text(query)
        )
        
        scores = self.bm25.get_scores(query_tokens)
        top_indices = np.argsort(scores)[::-1][:top_k]
        
        results = []
        for rank, idx in enumerate(top_indices, 1):
            results.append({
                'doc_id': self.documents[idx]['doc_id'],
                'text': self.documents[idx]['text'][:200],
                'score': float(scores[idx]),
                'metadata': self.documents[idx].get('metadata', {}),
                'rank': rank
            })
        
        return results
    
    def save(self, filepath):
        """Save BM25 index"""
        os.makedirs(os.path.dirname(filepath) or '.', exist_ok=True)
        with open(filepath, 'wb') as f:
            pickle.dump(self.bm25, f)
        print(f"✓ BM25 index saved to: {filepath}")

# Initialize BM25
bm25_retriever = BM25Retriever(documents, tokenized_docs)
print()

# Test BM25
print("Testing BM25 Retrieval:")
print("-" * 60)
test_query = "patient diabetes blood pressure"
bm25_results = bm25_retriever.search(test_query, top_k=3)

print(f"Query: {test_query}")
print(f"Found {len(bm25_results)} results\n")

for result in bm25_results:
    print(f"Rank {result['rank']}: {result['doc_id']}")
    print(f"Score: {result['score']:.4f}")
    print(f"Text: {result['text']}...")
    print()
```

```python
# ============================================
# Cell 4: Simple Dense Retriever (No Model)
# ============================================
print("=" * 60)
print("STEP 4: Building Simple Dense Retriever")
print("=" * 60)

class SimpleDenseRetriever:
    """Simple dense retriever using TF-IDF (no ML library needed)"""
    
    def __init__(self, documents, tokenized_docs):
        self.documents = documents
        self.tokenized_docs = tokenized_docs
        self.preprocessor = TextPreprocessor()
        
        print("Building TF-IDF index...")
        self.tfidf_vectors = self._build_tfidf()
        print("✓ TF-IDF index built successfully!")
    
    def _build_tfidf(self):
        """Build TF-IDF vectors for all documents"""
        # Build vocabulary
        vocab = set()
        for doc in self.tokenized_docs:
            vocab.update(doc)
        
        self.vocab = sorted(list(vocab))
        self.word_to_idx = {word: i for i, word in enumerate(self.vocab)}
        
        # Calculate IDF
        doc_count = len(self.tokenized_docs)
        idf = np.zeros(len(self.vocab))
        
        for word_idx, word in enumerate(self.vocab):
            doc_freq = sum(1 for doc in self.tokenized_docs if word in doc)
            idf[word_idx] = math.log(doc_count / (1 + doc_freq))
        
        self.idf = idf
        
        # Calculate TF-IDF vectors
        vectors = []
        for doc in self.tokenized_docs:
            vector = np.zeros(len(self.vocab))
            doc_length = len(doc)
            
            for word in doc:
                word_idx = self.word_to_idx[word]
                vector[word_idx] += 1
            
            # Normalize by document length and apply IDF
            if doc_length > 0:
                vector = (vector / doc_length) * self.idf
            
            vectors.append(vector)
        
        return np.array(vectors)
    
    def cosine_similarity(self, vec1, vec2):
        """Calculate cosine similarity"""
        dot_product = np.dot(vec1, vec2)
        magnitude1 = np.sqrt(np.sum(vec1 ** 2))
        magnitude2 = np.sqrt(np.sum(vec2 ** 2))
        
        if magnitude1 == 0 or magnitude2 == 0:
            return 0
        
        return dot_product / (magnitude1 * magnitude2)
    
    def search(self, query, top_k=5):
        """Search using TF-IDF"""
        query_tokens = self.preprocessor.tokenize(
            self.preprocessor.clean_text(query)
        )
        
        # Create query vector
        query_vector = np.zeros(len(self.vocab))
        for token in query_tokens:
            if token in self.word_to_idx:
                word_idx = self.word_to_idx[token]
                query_vector[word_idx] += 1
        
        # Normalize and apply IDF
        query_length = len(query_tokens)
        if query_length > 0:
            query_vector = (query_vector / query_length) * self.idf
        
        # Calculate similarities
        scores = np.array([
            self.cosine_similarity(query_vector, doc_vec)
            for doc_vec in self.tfidf_vectors
        ])
        
        # Get top-k
        top_indices = np.argsort(scores)[::-1][:top_k]
        
        results = []
        for rank, idx in enumerate(top_indices, 1):
            results.append({
                'doc_id': self.documents[idx]['doc_id'],
                'text': self.documents[idx]['text'][:200],
                'score': float(scores[idx]),
                'metadata': self.documents[idx].get('metadata', {}),
                'rank': rank
            })
        
        return results

# Initialize dense retriever
dense_retriever = SimpleDenseRetriever(documents, tokenized_docs)
print()

# Test dense retriever
print("Testing Dense Retrieval (TF-IDF):")
print("-" * 60)
dense_results = dense_retriever.search(test_query, top_k=3)

print(f"Query: {test_query}")
print(f"Found {len(dense_results)} results\n")

for result in dense_results:
    print(f"Rank {result['rank']}: {result['doc_id']}")
    print(f"Score: {result['score']:.4f}")
    print(f"Text: {result['text']}...")
    print()
```

```python
# ============================================
# Cell 5: Hybrid Retriever
# ============================================
print("=" * 60)
print("STEP 5: Building Hybrid Retriever")
print("=" * 60)

class HybridRetriever:
    """Hybrid retriever combining BM25 and TF-IDF"""
    
    def __init__(self, bm25_retriever, dense_retriever, alpha=0.5):
        self.bm25_retriever = bm25_retriever
        self.dense_retriever = dense_retriever
        self.alpha = alpha
    
    def normalize_scores(self, results):
        """Normalize scores to 0-1 range"""
        if not results:
            return results
        
        scores = [r['score'] for r in results]
        min_score = min(scores)
        max_score = max(scores)
        
        if max_score - min_score == 0:
            for r in results:
                r['normalized_score'] = 1.0
        else:
            for r in results:
                r['normalized_score'] = (r['score'] - min_score) / (max_score - min_score)
        
        return results
    
    def search(self, query, top_k=5, bm25_k=10, dense_k=10):
        """Hybrid search"""
        bm25_results = self.bm25_retriever.search(query, top_k=bm25_k)
        bm25_results = self.normalize_scores(bm25_results)
        
        dense_results = self.dense_retriever.search(query, top_k=dense_k)
        dense_results = self.normalize_scores(dense_results)
        
        doc_scores = {}
        doc_data = {}
        
        for result in bm25_results:
            doc_id = result['doc_id']
            doc_scores[doc_id] = self.alpha * result['normalized_score']
            doc_data[doc_id] = result
        
        for result in dense_results:
            doc_id = result['doc_id']
            if doc_id in doc_scores:
                doc_scores[doc_id] += (1 - self.alpha) * result['normalized_score']
            else:
                doc_scores[doc_id] = (1 - self.alpha) * result['normalized_score']
                doc_data[doc_id] = result
        
        sorted_docs = sorted(doc_scores.items(), key=lambda x: x[1], reverse=True)
        
        final_results = []
        for rank, (doc_id, score) in enumerate(sorted_docs[:top_k], 1):
            result = doc_data[doc_id].copy()
            result['hybrid_score'] = score
            result['rank'] = rank
            final_results.append(result)
        
        return final_results

# Initialize hybrid
hybrid_retriever = HybridRetriever(bm25_retriever, dense_retriever, alpha=0.5)
print("✓ Hybrid retriever initialized!")
print()

# Test hybrid
print("Testing Hybrid Retrieval:")
print("-" * 60)
hybrid_results = hybrid_retriever.search(test_query, top_k=3)

print(f"Query: {test_query}")
print(f"Found {len(hybrid_results)} results\n")

for result in hybrid_results:
    print(f"Rank {result['rank']}: {result['doc_id']}")
    print(f"Hybrid Score: {result['hybrid_score']:.4f}")
    print(f"Text: {result['text']}...")
    print()
```

```python
# ============================================
# SIMPLE LLM WRAPPER (Fallback)
# ============================================

class SimpleLLM:
    """Simple template-based LLM for when models unavailable"""
    
    def __init__(self):
        self.responses = {
            'diagnosis': "Based on the clinical information provided, the patient presents with {condition}. Further evaluation and testing may be required to confirm the diagnosis.",
            'treatment': "Treatment recommendations should include {treatment}. Close monitoring and follow-up appointments are essential.",
            'prognosis': "The prognosis depends on multiple factors including patient compliance with treatment and overall health status.",
            'general': "According to the clinical documentation provided, {answer}"
        }
    
    def generate(self, prompt, context="", max_length=200):
        """Generate response using templates"""
        # Extract key terms
        prompt_lower = prompt.lower()
        
        if any(word in prompt_lower for word in ['diagnosis', 'diagnose', 'what is']):
            template = self.responses['diagnosis']
            condition = self._extract_condition(context)
            return template.format(condition=condition or "multiple conditions")
        
        elif any(word in prompt_lower for word in ['treatment', 'treat', 'medicine', 'medication']):
            template = self.responses['treatment']
            treatment = self._extract_treatment(context)
            return template.format(treatment=treatment or "appropriate medical intervention")
        
        elif any(word in prompt_lower for word in ['prognosis', 'outcome', 'future']):
            return self.responses['prognosis']
        
        else:
            template = self.responses['general']
            summary = self._summarize_context(context)
            return template.format(answer=summary)
    
    def _extract_condition(self, context):
        """Extract condition from context"""
        keywords = ['diabetes', 'hypertension', 'fever', 'infection', 'pain', 'cardiac']
        for keyword in keywords:
            if keyword in context.lower():
                return keyword
        return None
    
    def _extract_treatment(self, context):
        """Extract treatment info"""
        keywords = ['medication', 'insulin', 'therapy', 'treatment', 'antibiotic']
        for keyword in keywords:
            if keyword in context.lower():
                return keyword
        return None
    
    def _summarize_context(self, context):
        """Summarize context"""
        sentences = context.split('.')
        return sentences[0][:100] if sentences else "Clinical information available"
# ============================================
# ADVANCED LLM MODEL (Hugging Face)
# ============================================

class LLMGenerator:
    """Advanced LLM Generator using Hugging Face models"""
    
    def __init__(self, model_name='gpt2', use_gpu=False):
        """
        Initialize LLM
        model_name: Model to use (gpt2, distilgpt2, etc.)
        use_gpu: Whether to use GPU if available
        """
        self.model_name = model_name
        self.use_gpu = use_gpu and torch.cuda.is_available()
        self.device = "cuda" if self.use_gpu else "cpu"
        
        print(f"Loading model: {model_name}")
        print(f"Device: {self.device}")
        
        try:
            # Load tokenizer and model
            self.tokenizer = AutoTokenizer.from_pretrained(model_name)
            self.model = AutoModelForCausalLM.from_pretrained(model_name)
            self.model = self.model.to(self.device)
            
            # Set pad token
            if self.tokenizer.pad_token is None:
                self.tokenizer.pad_token = self.tokenizer.eos_token
            
            print(f"✓ Model loaded successfully!")
            self.model_available = True
        except Exception as e:
            print(f"⚠️ Model loading failed: {e}")
            print("Using fallback template-based LLM")
            self.model_available = False
            self.fallback_llm = SimpleLLM()
    
    def generate(self, prompt, context="", max_length=200, temperature=0.7):
        """Generate response"""
        
        if not self.model_available:
            return self.fallback_llm.generate(prompt, context, max_length)
        
        try:
            # Prepare input
            combined_input = f"Context: {context}\n\nQuestion: {prompt}\n\nAnswer:"
            
            inputs = self.tokenizer(
                combined_input,
                return_tensors="pt",
                truncation=True,
                max_length=512
            ).to(self.device)
            
            # Generate
            with torch.no_grad():
                outputs = self.model.generate(
                    **inputs,
                    max_length=max_length,
                    temperature=temperature,
                    top_p=0.9,
                    do_sample=True,
                    pad_token_id=self.tokenizer.eos_token_id
                )
            
            # Decode
            response = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
            
            # Extract answer part
            if "Answer:" in response:
                response = response.split("Answer:")[-1].strip()
            
            return response.strip()
        
        except Exception as e:
            print(f"⚠️ Generation failed: {e}")
            return self.fallback_llm.generate(prompt, context, max_length)
# ============================================
# PROMPT ENGINEERING
# ============================================

class PromptEngineer:
    """Advanced prompt engineering for clinical context"""
    
    def __init__(self):
        self.templates = {
            'diagnosis': """You are an expert clinical assistant. Based on the following clinical documents, answer the question about diagnosis.

Clinical Context:
{context}

Question: {query}

Please provide a detailed response based on the clinical information provided:""",
            
            'treatment': """You are an expert clinical assistant specializing in treatment recommendations. Based on the clinical documents provided, answer the question about treatment options.

Clinical Context:
{context}

Question: {query}

Please provide evidence-based treatment recommendations:""",
            
            'prognosis': """You are an expert clinical assistant. Based on the following clinical information, answer the question about prognosis and outcomes.

Clinical Context:
{context}

Question: {query}

Please provide an analysis of the likely prognosis:""",
            
            'general': """You are an expert clinical assistant. Based on the clinical documents provided, answer the following question accurately and comprehensively.

Clinical Context:
{context}

Question: {query}

Answer:""",
            
            'comparison': """You are an expert clinical assistant. Compare the following clinical information and answer the question.

Clinical Context:
{context}

Question: {query}

Please provide a detailed comparison:""",
        }
        
        self.keywords = {
            'diagnosis': ['diagnose', 'diagnosis', 'condition', 'disease', 'what is', 'identify'],
            'treatment': ['treat', 'treatment', 'medicine', 'medication', 'therapy', 'how to treat'],
            'prognosis': ['prognosis', 'outcome', 'future', 'recovery', 'will', 'likely'],
            'comparison': ['compare', 'difference', 'vs', 'versus', 'similar', 'different'],
        }
    
    def detect_query_type(self, query):
        """Detect the type of query"""
        query_lower = query.lower()
        
        for query_type, keywords in self.keywords.items():
            for keyword in keywords:
                if keyword in query_lower:
                    return query_type
        
        return 'general'
    
    def engineer_prompt(self, query, context, retrieved_docs):
        """Engineer prompt with retrieved context"""
        
        # Detect query type
        query_type = self.detect_query_type(query)
        template = self.templates.get(query_type, self.templates['general'])
        
        # Prepare context from retrieved documents
        context_text = self._prepare_context(retrieved_docs)
        
        # Build prompt
        prompt = template.format(
            context=context_text,
            query=query
        )
        
        return prompt, query_type
    
    def _prepare_context(self, retrieved_docs):
        """Prepare context from retrieved documents"""
        context_parts = []
        
        for doc in retrieved_docs:
            doc_text = f"[Doc: {doc.get('doc_id', 'Unknown')}]\n"
            doc_text += doc.get('text', '')[:300]  # Limit to 300 chars per doc
            doc_text += f"\n[Relevance Score: {doc.get('score', 0):.2f}]\n"
            context_parts.append(doc_text)
        
        return "\n---\n".join(context_parts)
# ============================================
# COMPLETE RAG PIPELINE
# ============================================

class RAGPipeline:
    """Complete Retrieval-Augmented Generation Pipeline"""
    
    def __init__(self, retrieval_pipeline, llm_generator=None):
        """
        Initialize RAG Pipeline
        retrieval_pipeline: Retrieval system from Task 3
        llm_generator: LLM for generation (optional)
        """
        self.retrieval_pipeline = retrieval_pipeline
        
        # Initialize LLM
        if llm_generator is None:
            print("Initializing LLM Generator...")
            try:
                self.llm = LLMGenerator(model_name='distilgpt2', use_gpu=False)
            except:
                print("⚠️ Using fallback template-based LLM")
                self.llm = SimpleLLM()
        else:
            self.llm = llm_generator
        
        # Initialize prompt engineer
        self.prompt_engineer = PromptEngineer()
        
        # Conversation history
        self.conversation_history = []
    
    def process_query(self, query, retrieval_method='hybrid', top_k=5):
        """
        Process a clinical query through complete RAG pipeline
        
        Args:
            query: Clinical question
            retrieval_method: 'hybrid', 'bm25', or 'dense'
            top_k: Number of documents to retrieve
        
        Returns:
            Dictionary with query, retrieved docs, and generated answer
        """
        
        # Step 1: Retrieve relevant documents
        print("\n" + "=" * 70)
        print(f"Processing Query: {query}")
        print("=" * 70)
        
        print("\n[RETRIEVAL PHASE]")
        retrieved_docs = self.retrieval_pipeline.retrieve(
            query,
            method=retrieval_method,
            top_k=top_k
        )
        
        print(f"✓ Retrieved {len(retrieved_docs)} relevant documents")
        for doc in retrieved_docs:
            print(f"  - {doc['doc_id']}: Score {doc.get('hybrid_score', doc.get('score', 0)):.4f}")
        
        # Step 2: Engineer prompt
        print("\n[PROMPT ENGINEERING PHASE]")
        prompt, query_type = self.prompt_engineer.engineer_prompt(
            query,
            "",
            retrieved_docs
        )
        
        print(f"✓ Query Type Detected: {query_type}")
        print(f"✓ Prompt engineered with {len(retrieved_docs)} documents")
        
        # Step 3: Generate response
        print("\n[GENERATION PHASE]")
        print("Generating response...")
        
        context_text = self.prompt_engineer._prepare_context(retrieved_docs)
        generated_answer = self.llm.generate(
            prompt=query,
            context=context_text,
            max_length=250
        )
        
        print(f"✓ Response generated successfully!")
        
        # Step 4: Prepare result
        result = {
            'timestamp': datetime.now().isoformat(),
            'query': query,
            'query_type': query_type,
            'retrieval_method': retrieval_method,
            'retrieved_documents': retrieved_docs,
            'generated_answer': generated_answer,
            'num_retrieved_docs': len(retrieved_docs),
            'top_doc_score': retrieved_docs[0].get('score', 0) if retrieved_docs else 0
        }
        
        # Add to conversation history
        self.conversation_history.append(result)
        
        return result
    
    def display_result(self, result):
        """Display query result beautifully"""
        
        print("\n" + "=" * 70)
        print("FINAL RESULT")
        print("=" * 70)
        
        print(f"\n📋 Query: {result['query']}")
        print(f"🏷️  Query Type: {result['query_type']}")
        print(f"🔍 Retrieval Method: {result['retrieval_method']}")
        
        print(f"\n📚 Retrieved Documents ({result['num_retrieved_docs']}):")
        print("-" * 70)
        for doc in result['retrieved_documents']:
            print(f"\n{doc['doc_id']} (Score: {doc.get('score', 0):.4f})")
            print(f"Text: {doc['text']}...")
        
        print(f"\n🤖 Generated Answer:")
        print("-" * 70)
        print(result['generated_answer'])
        
        print("\n" + "=" * 70)
    
    def batch_process(self, queries, retrieval_method='hybrid'):
        """Process multiple queries"""
        results = []
        for query in queries:
            result = self.process_query(query, retrieval_method)
            results.append(result)
        return results
    
    def save_results(self, filepath):
        """Save conversation history"""
        os.makedirs(os.path.dirname(filepath) or '.', exist_ok=True)
        with open(filepath, 'w') as f:
            json.dump(self.conversation_history, f, indent=2, default=str)
        print(f"✓ Results saved to: {filepath}")
```

```python
# ============================================
# LOAD TASK 3 RETRIEVAL SYSTEM
# ============================================
print("=" * 70)
print("STEP 2: Loading Retrieval System from Task 3")
print("=" * 70)

# Load preprocessed documents
preprocessed_path = "preprocessed_data/preprocessed_documents.json"
with open(preprocessed_path, 'r') as f:
    documents = json.load(f)

print(f"✓ Loaded {len(documents)} documents")

# Load retrieval system (simplified version for this demo)
from collections import Counter
import math

class MinimalRetrieval:
    """Minimal retrieval for demo (if Task 3 components not available)"""
    
    def __init__(self, documents):
        self.documents = documents
    
    def retrieve(self, query, method='hybrid', top_k=5):
        """Simple keyword matching retrieval"""
        query_words = set(query.lower().split())
        
        scores = []
        for doc in self.documents:
            doc_text = doc['text'].lower()
            # Count matching words
            matches = sum(1 for word in query_words if word in doc_text)
            scores.append((doc, matches))
        
        # Sort by score and return top-k
        scores.sort(key=lambda x: x[1], reverse=True)
        top_docs = scores[:top_k]
        
        results = []
        for rank, (doc, score) in enumerate(top_docs, 1):
            results.append({
                'doc_id': doc['doc_id'],
                'text': doc['text'][:200],
                'score': float(score) / (len(query_words) + 1),
                'hybrid_score': float(score) / (len(query_words) + 1),
                'metadata': doc.get('metadata', {}),
                'rank': rank
            })
        
        return results

retrieval_system = MinimalRetrieval(documents)
print("✓ Retrieval system loaded!")
print()

# ============================================
# INITIALIZE RAG PIPELINE
# ============================================
print("=" * 70)
print("STEP 3: Initializing RAG Pipeline")
print("=" * 70)

rag_pipeline = RAGPipeline(retrieval_system)
print("✓ RAG Pipeline initialized!")
print()
```
